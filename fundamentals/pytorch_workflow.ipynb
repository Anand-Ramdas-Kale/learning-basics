### machine learning is a game of two parts:
* Get the data into numeracal representation 
* Build model to learn patterns in that numerical representation

```python
import torch
import matplotlib.pyplot as plt
```

known parameters **baby model**

```python
weight = 0.7
bias = 0.3

start = 0.0
end = 1
step = 0.02

x = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * x + bias
```

```python
print(f"x = {x}")
print("\n")
print(f"y = {y}")
```

### splitting data into train and test data
* training set
* validation set
* test set

```python
train_split = int(0.8 * len(x))
x_train, y_train = x[:train_split], y[:train_split]
x_test, y_test = x[train_split:], y[train_split:]
```

```python
len(x_train), len(y_train), len(x_test), len(y_test)
```

```python
def plot_predictions(train_data=x_train,
                    train_labels=y_train,
                    test_data=x_test,
                    test_labels=y_test,
                    predictions=None):
    """
    plots training data and test data and the predictions
    """
    plt.figure(figsize=(10,9))
    plt.scatter(train_data, train_labels, c="b", s=4, label="training data")
    
plot_predictions()
```

```python
def plot_predictions(train_data=x_train,
                    train_labels=y_train,
                    test_data=x_test,
                    test_labels=y_test,
                    predictions=None):
    """
    plots training data and test data and the predictions
    """
    plt.figure(figsize=(10,9))
    plt.scatter(train_data, train_labels, c="b", s=4, label="training data")
    plt.scatter(test_data, test_labels, c="g", s=4, label="testing data")

    if predictions is not None:
        plt.scatter(test_data, predictions, c="r", s=4, label="predictions")
    plt.legend(prop={"size":12})
    
plot_predictions()
```

```python
from torch import nn

class LinearRegressionModel(nn.Module):
    def __init__(self):
        super.__init__()
        self.weights = nn.Parameter(torch.randn(1,
                                               requires_grad=True,
                                               dtype=torch.float))
        self.bias = nn.Parameter(torch.randn(1,
                                               requires_grad=True,
                                               dtype=torch.float))

    def forward(self, x:torch.Tensor) -> torch.Tensor:
        return self.weights * x + self.bias
```

* torch.nn - contains all the building blocks for computational graphs (neural networks)
* torch.nn.Parameters - what parameters should our model try and learn, often a pytorch layer from torch.nn will set this for us
* torch.nn.Module - the base class for all neural network modules
* torch.optim - this is where the optimizers in pytorch live, they will help in gradient descent
