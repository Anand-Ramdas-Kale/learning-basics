# matplotlib basics

```python
from matplotlib import pyplot as plt
import numpy as np
```

np arrawys are faster.

```python
x = np.array([2023, 2024, 2025, 2026])
y0 = np.array([15, 25, 30, 20])
y1 = np.array([20, 10, 50, 5])
y2 = np.array([10, 15, 20, 30])

plt.plot(x, y0)
plt.show()
```

we can have line style stored in a dictionary and pass the unpacked dict to plot

```python
line_style = dict(
    marker=".",
    markersize="20",
    markerfacecolor="#03f8fc",
    markeredgecolor="#1d5354",
    linestyle="dashed",
    linewidth="3",
)
```

plotting multiple line graphs

```python
plt.title("this is the title", fontsize=18, fontweight="bold", color="#f74569")
plt.xlabel("xlabel", fontsize=15)
# `family` for font-family

plt.plot(x,y0,**line_style, color="red")
plt.plot(x,y1,**line_style, color="blue")
plt.plot(x,y2,**line_style, color="orange")
plt.show()
```

## tick params

```python
plt.tick_params(axis="both",
               colors="#0581f5")
plt.xticks(x)

plt.title("this is the title", fontsize=18, fontweight="bold", color="#f74569")
plt.xlabel("xlabel", fontsize=15)
plt.ylabel("ylabel", fontsize=15)
# `family` for font-family

plt.plot(x,y0,**line_style, color="red")
plt.plot(x,y1,**line_style, color="blue")
plt.plot(x,y2,**line_style, color="orange")
plt.show()
```

## Subplot adjust
makes use of same canvas to plot different plots on different sizes 

```python
plt.subplots_adjust(bottom=0.3,left=0.2)

plt.tick_params(axis="both",
               colors="#0581f5")
plt.xticks(x)

plt.title("this is the title", fontsize=18, fontweight="bold", color="#f74569")
plt.xlabel("xlabel", fontsize=15)
plt.ylabel("ylabel", fontsize=15)
# `family` for font-family

plt.plot(x,y0,**line_style, color="red")
plt.plot(x,y1,**line_style, color="blue")
plt.plot(x,y2,**line_style, color="orange")
plt.show()
```

## grid

```python
xg = np.arange(-5,5,0.5)
yg = xg**2
plt.plot(xg,yg)
plt.grid()
plt.show()
```

we can show single axis grids also

```python
xg = np.arange(-5,5,0.5)
yg = xg**2
plt.plot(xg,yg)
plt.grid(axis="x",
        linewidth=2,
        color="#84618c",
        linestyle="dashdot")
plt.show()
```

## BAR graph

```python
categories = ["Grains", "Fruits", "Vegetables", "Proteins", "Dairy", "Sweets"]
values = [4,3,2,6,3,1]

plt.bar(categories, values, color="skyblue")

plt.show()
plt.barh(categories, values, color="skyblue")

plt.show()
```

## Pie Charts

```python
categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = np.array([200, 500, 329, 102])
colors = ["red", "yellow", "blue", "green"]

plt.pie(values, 
        labels=categories, 
        autopct="%1.2f%%", 
        colors=colors, 
        explode=[0,0,0,0.2],
        shadow="True",
        startangle=90
       )
plt.show()
```
