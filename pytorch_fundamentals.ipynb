```python
import torch
```

```python
x = torch.arange(1., 10.)
x, x.shape
```

```python
x_reshaped = x.reshape(3,3)
x_reshaped, x_reshaped.shape
```

```python
x_reshaped_2 = x.reshape(1,9)
x_reshaped_2
```

### view
* a view shares the memory with the original tensor
* transpose x.T also has the same memory as x

```python
z = x.view(3,3)
z, x
```

```python
x_stacked = torch.stack([x,x,x], dim = 1)
x_stacked, x_stacked.shape
```

```python
x_vstacked = torch.vstack([x,x,x])
x_vstacked
```

```python
x_hstacked = torch.hstack([x,x,x])
x_hstacked
```

### squeeze()
* removes all single dimension

```python
x_squeezed = x_reshaped_2.squeeze()
print(f"x_squeezed = {x_squeezed}")
print(f"x_squeezed.shape = {x_squeezed.shape}")
```

### unsqueeze(dim=n) 
* adds single dimension at specified dimension


```python
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(f"x_unsqueezed = {x_unsqueezed}")
print(f"x_unsqueezed.shape = {x_unsqueezed.shape}")
```

### permute()
* rearranges the dimensions of tensor in specified order
* so changes the order of co-ordinates of the points present -- for understanding purposes
* shares the same memory with the tensor, just a different view

```python
f = torch.rand(5,4,3)
f
```

```python
f_permuted = f.permute(2,0,1)
f_permuted
```

```python
f_one = f.reshape(-1)
f_one
```

```python
f_permuted_one = f_permuted.reshape(-1)
f_permuted_one
```

```python
f[0,0,0] = 1
f, f_permuted
```

### back and forth between numpy and torch
* torch.from_numpy()
* torch.Tensor.numpy()
* have different memory
* torch preserves the datatype of numpy as f64, which is not torch default (f32)
* going from torch to numpy preserves it as f32

```python
import torch
import numpy as np
```

```python
array = np.arange(1,8)
tensor = torch.from_numpy(array).type(torch.float32)
array, tensor
```

```python
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor.dtype, numpy_tensor.dtype
```

### Reproducibility
* **random seed**
* works per **use** basis


```python
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
A = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
B = torch.rand(3, 4)

C = torch.rand(3, 4)

print(A == B)
print("\n")
print(A == C)
```
