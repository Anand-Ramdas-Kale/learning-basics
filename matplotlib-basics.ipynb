## Markers

```python
# =============================================================================================================== #
#                                                FORMATTING CELL                                                  #
# =============================================================================================================== #
RESET = "\033[0m"
BOLD = "\033[1m"
DIM = "\033[2m"
ITALIC = "\033[3m"
UNDERLINE = "\033[4m"
HIGHLIGHT = "\033[7m"
STRIKE = "\033[9m"

RED = "\033[31m"
GREEN = "\033[32m"
YELLOW = "\033[33m"
BLUE = "\033[34m"
MAGENTA = "\033[35m"
CYAN = "\033[36m"

print(f"{BOLD}{RED}ERROR:{RESET} Something went wrong")
print(f"{GREEN}Success!{RESET}")

print("\033[1mBold\033[0m")
print("\033[2mDim\033[0m")
print("\033[3mItalic\033[0m")
print("\033[4mUnderline\033[0m")
print("\033[9mStrikethrough\033[0m")

print("\033[41mRed background\033[0m")
print("\033[42mGreen background\033[0m")
print("\033[44mBlue background\033[0m")
print("\033[45mMagenta background\033[0m")

print("\033[38;2;255;100;50mOrange text\033[0m")
for i in range(256):
    # the 5 is for 256 color palette, 3 in 38 is for forground, can do 4 for background
    print(f"\033[38;5;{i}m color - {i}\033[0m")

def print_separator():
    print(f"{RED}================================================================================{RESET}")
```
